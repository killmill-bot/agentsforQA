```python
import asyncio
import subprocess

async def test_subprocess():
    process = await asyncio.create_subprocess_exec(
        "cmd.exe",
        "/c",
        "echo",
        "SUBPROCESS_OK",
        stdout=asyncio.subprocess.PIPE
    )

    stdout, _ = await process.communicate()
    print(stdout.decode().strip())

await test_subprocess()
```

```python
import asyncio
import threading

def test_proactor():
    async def run():
        print("Loop:", type(asyncio.get_running_loop()))

        process = await asyncio.create_subprocess_exec(
            "cmd.exe",
            "/c",
            "echo",
            "PROACTOR_OK",
            stdout=asyncio.subprocess.PIPE
        )

        stdout, _ = await process.communicate()
        print(stdout.decode().strip())

    loop = asyncio.ProactorEventLoop()

    try:
        loop.run_until_complete(run())
    finally:
        loop.close()


thread = threading.Thread(target=test_proactor)
thread.start()
thread.join()
```

```python
import asyncio
import threading
import traceback

def run_playwright():
    async def test():
        print("1. test gestart", flush=True)

        from playwright.async_api import async_playwright
        print("2. Playwright geïmporteerd", flush=True)

        print(
            "3. Loop:",
            type(asyncio.get_running_loop()),
            flush=True
        )

        async with async_playwright() as p:
            print("4. Playwright gestart", flush=True)

            browser = await p.chromium.launch(headless=True)
            print("5. Chromium gestart", flush=True)

            page = await browser.new_page()
            print("6. Page aangemaakt", flush=True)

            await page.goto("https://example.com")
            print("7. Pagina geladen", flush=True)

            print("Titel:", await page.title(), flush=True)
            print("URL:", page.url, flush=True)

            await browser.close()
            print("8. Browser gesloten", flush=True)

    loop = asyncio.ProactorEventLoop()

    try:
        asyncio.set_event_loop(loop)
        print("Loop gestart", flush=True)
        loop.run_until_complete(test())
        print("Test afgerond", flush=True)

    except Exception:
        traceback.print_exc()

    finally:
        loop.close()
        print("Loop gesloten", flush=True)


thread = threading.Thread(target=run_playwright)
thread.start()
thread.join()
```

```python
import asyncio
import threading

def run_playwright(coroutine):
    result = []
    error = []

    def worker():
        loop = asyncio.ProactorEventLoop()
        asyncio.set_event_loop(loop)

        try:
            result.append(loop.run_until_complete(coroutine()))
        except Exception as e:
            error.append(e)
        finally:
            loop.close()

    thread = threading.Thread(target=worker)
    thread.start()
    thread.join()

    if error:
        raise error[0]

    return result[0] if result else None
```
